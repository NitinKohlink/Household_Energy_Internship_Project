# 01 - Data Collection & Quality Check

I started by checking what is actually inside the supplied CSV before doing any modeling.

```python
import pandas as pd

file = "../01_Data/raw/household_power_consumption.csv"
df = pd.read_csv(file, na_values="?")
df["datetime"] = pd.to_datetime(df["Date"] + " " + df["Time"], dayfirst=True)

print("Rows:", len(df))
print("Columns:", df.shape[1])
print("Start:", df["datetime"].min())
print("End:", df["datetime"].max())
```

## Missing values and duplicates
I checked missing values, duplicate rows and duplicate timestamps. The goal was to know the data quality before changing anything.

```python
print(df.isna().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate timestamps:", df["datetime"].duplicated().sum())
```

```python
gap = df["datetime"].diff().dt.total_seconds().div(60)
print("1-minute intervals (%):", round(gap.dropna().eq(1).mean()*100, 2))
print("Largest gap (minutes):", int(gap.max()))
```
