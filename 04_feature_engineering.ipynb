```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
hourly = pd.read_csv('../01_Data/processed/hourly_energy.csv', parse_dates=['date_time'])
```

```python
hourly.head()
```

```python
hourly = hourly.sort_values("date_time")
```

## Time features

```python
hourly["hour"] = hourly["date_time"].dt.hour
```

```python
hourly["dayofweek"] = hourly["date_time"].dt.dayofweek
```

```python
hourly["month"] = hourly["date_time"].dt.month
```

```python
hourly["is_weekend"] = (hourly["dayofweek"] >= 5).astype(int)
```

## Lag features

```python
hourly["lag_1"] = hourly["energy_kwh"].shift(1)
```

```python
hourly["lag_24"] = hourly["energy_kwh"].shift(24)
```

```python
hourly["lag_168"] = hourly["energy_kwh"].shift(168)
```

```python
hourly[["energy_kwh","lag_1","lag_24","lag_168"]].head(175)
```

## Rolling features

```python
hourly["rolling_24"] = hourly["energy_kwh"].shift(1).rolling(24).mean()
```

```python
hourly["rolling_168"] = hourly["energy_kwh"].shift(1).rolling(168).mean()
```

```python
hourly[["energy_kwh","rolling_24","rolling_168"]].head(175)
```

## Removing first missing rows

```python
hourly = hourly.dropna()
```

```python
hourly.shape
```

## Comparing actual and previous hour

```python
plt.figure(figsize=(10,4))
plt.plot(hourly["date_time"].tail(168), hourly["energy_kwh"].tail(168), label="Actual")
plt.plot(hourly["date_time"].tail(168), hourly["lag_1"].tail(168), label="Lag 1")
plt.legend()
plt.title("Energy and Previous Hour")
plt.show()
```

```python
plt.savefig('../04_Visuals/ml/07_feature_engineering_check.png', dpi=150)
plt.show()
```

## Selecting features

```python
features = ["hour","dayofweek","month","is_weekend","lag_1","lag_24","lag_168","rolling_24","rolling_168"]
```

```python
hourly[["date_time","energy_kwh"] + features].head()
```

```python
model_data = hourly[["date_time","energy_kwh"] + features]
```
