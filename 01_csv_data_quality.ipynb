## Checking missing values

```python
df.isna().sum()
```

```python
df.isna().sum().sum()
```

```python
missing_rows = df.isna().any(axis=1).sum()
print("Rows with missing values", missing_rows)
```

```python
print("Missing rows %", round(missing_rows / len(df) * 100, 2))
```

## Checking duplicates

```python
df.duplicated().sum()
```

## Checking numeric columns

```python
num_cols = ['Global_active_power','Global_reactive_power','Voltage','Global_intensity','Sub_metering_1','Sub_metering_2','Sub_metering_3']
```

```python
for col in num_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
```

```python
df[num_cols].describe().round(2)
```

## Missing values graph

```python
missing = df.isna().sum()
missing = missing[missing > 0]
```

```python
missing
```

```python
plt.figure(figsize=(8,4))
missing.plot(kind="bar")
plt.ylabel("Missing values")
plt.title("Missing Values")
plt.tight_layout()
plt.show()
```
