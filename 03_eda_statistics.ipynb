```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
```

```python
daily = pd.read_csv('../01_Data/processed/daily_energy.csv', parse_dates=['date'])
```

```python
hourly = pd.read_csv('../01_Data/processed/hourly_energy.csv', parse_dates=['date_time'])
```

```python
daily.head()
```

## Basic statistics

```python
daily["daily_kwh"].describe().round(2)
```

```python
print("Mean", round(daily["daily_kwh"].mean(), 2))
```

```python
print("Median", round(daily["daily_kwh"].median(), 2))
```

## Distribution

```python
plt.figure(figsize=(10,4))
plt.hist(daily["daily_kwh"], bins=30)
plt.xlabel("Daily energy (kWh)")
plt.title("Daily Energy Distribution")
plt.show()
```

```python
plt.savefig('../04_Visuals/eda/05_daily_energy_boxplot.png', dpi=150)
plt.show()
```

```python
plt.boxplot(daily["daily_kwh"])
plt.title("Daily Energy Boxplot")
plt.ylabel("Daily energy (kWh)")
plt.show()
```

## Daily trend

```python
plt.figure(figsize=(10,4))
plt.plot(daily["date"], daily["daily_kwh"])
plt.xlabel("Date")
plt.ylabel("kWh")
plt.title("Daily Energy Trend")
plt.show()
```

```python
plt.savefig('../04_Visuals/eda/01_daily_energy_trend.png', dpi=150)
plt.show()
```

## Hourly pattern

```python
hour_avg = hourly.groupby("hour")["energy_kwh"].mean()
```

```python
plt.figure(figsize=(9,4))
plt.plot(hour_avg.index, hour_avg.values, marker="o")
plt.xlabel("Hour")
plt.ylabel("Average kWh")
plt.title("Average Energy by Hour")
plt.xticks(range(0,24,2))
plt.show()
```

```python
plt.savefig('../04_Visuals/eda/02_hourly_profile.png', dpi=150)
plt.show()
```

## Weekday vs weekend

```python
weekday = daily[daily["is_weekend"] == 0]["daily_kwh"]
weekend = daily[daily["is_weekend"] == 1]["daily_kwh"]
```

```python
print("Weekday mean", round(weekday.mean(), 2))
print("Weekend mean", round(weekend.mean(), 2))
```

```python
uplift = (weekend.mean() / weekday.mean() - 1) * 100
print("Weekend uplift", round(uplift, 2), "%")
```

```python
plt.boxplot([weekday, weekend], labels=["Weekday", "Weekend"])
plt.ylabel("Daily energy (kWh)")
plt.title("Weekday vs Weekend")
plt.show()
```

```python
plt.savefig('../04_Visuals/eda/03_weekday_weekend.png', dpi=150)
plt.show()
```

## Correlation

```python
cols = ['daily_kwh','avg_voltage','avg_intensity','avg_reactive_power']
```

```python
daily[cols].corr().round(2)
```

```python
plt.figure(figsize=(7,5))
plt.imshow(daily[cols].corr(), cmap="coolwarm", vmin=-1, vmax=1)
plt.xticks(range(len(cols)), cols, rotation=45)
plt.yticks(range(len(cols)), cols)
plt.colorbar()
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.savefig('../04_Visuals/eda/04_correlation_heatmap.png', dpi=150)
plt.show()
```

## Simple statistical test

```python
t_stat, p_value = stats.ttest_ind(weekday, weekend, equal_var=False)
```

```python
print("t-stat", round(t_stat, 2))
print("p-value", round(p_value, 4))
```
